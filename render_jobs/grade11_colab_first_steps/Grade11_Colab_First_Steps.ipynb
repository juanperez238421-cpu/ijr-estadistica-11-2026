## 6. Basic operations


```python
a = 18
b = 7

print(a + b)      # addition
print(a - b)      # subtraction
print(a * b)      # multiplication
print(a / b)      # decimal division
print(a // b)     # integer division
print(a % b)      # remainder
print(a ** 2)     # power
```

## 8. Comparisons

Comparisons return `True` or `False`.


```python
score = 78

print(score >= 70)
print(score < 60)
print(score == 78)
print(score != 80)
```

## 9. Boolean logic


```python
score = 78
attendance = 92

ready = score >= 70 and attendance >= 85
support = score < 60 or attendance < 75
not_ready = not ready

print(ready, support, not_ready)
```

## 10. Conditional logic

Python checks `if`, then `elif`, then `else`.


```python
score = 78

if score >= 80:
    level = "High"
elif score >= 60:
    level = "Developing"
else:
    level = "Support"

print(level)
```
